```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\AI-Data-Science-Career-Intelligence\data\raw\ai_ds_job_salaries_2026.csv")
```

```python
#Data Science and AI distribution
plt.figure(figsize=(10,6))
sns.histplot(df["salary_usd"], bins=30, kde=True)
plt.title("Distribution of Data Science & AI Salaries")
plt.xlabel("Salary_usd")
plt.ylabel("Number of Professionals")

plt.show()
```

Salary Distribution

The histogram shows that salaries are right-skewed.

Most professionals are concentrated roughly between $40,000 and $120,000, while there is a long tail extending toward $300,000+.

That explains something we saw earlier:

Mean salary   = $98,605
Median salary = $90,572

Because the mean is higher than the median, the higher-paid professionals are pulling the average upward.

What we can say in the project

Salary distribution is positively skewed, with most professionals concentrated in the lower-to-middle salary range and a smaller number of high earners extending beyond $300,000.

This is important because it tells us that mean salary alone doesn't tell the whole story. We'll use the median and distributions as well.

```python
salary_by_experience = (
    df.groupby("experience_level")["salary_usd"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(x=salary_by_experience.index, y=salary_by_experience.values)

plt.title("Average Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Average Salary (USD)")

plt.show()
```

There is a very clear upward relationship between career level and average salary:

Higher experience levels are associated with substantially higher average salaries in this dataset.

This is one of the strongest charts in our project, so we'll definitely use it later in the Power BI dashboard.
