```python
#Data Science and AI distribution
plt.figure(figsize=(10,6))
sns.histplot(df["salary_usd"], bins=30, kde=True)
plt.title("Distribution of Data Science & AI Salaries")
plt.xlabel("Salary_usd")
plt.ylabel("Number of Professionals")

plt.show()
```

# Salary Distribution

The histogram shows that salaries are right-skewed.

Most professionals are concentrated roughly between $40,000 and $120,000, while there is a long tail extending toward $300,000+.

That explains something we saw earlier:

Mean salary   = $98,605
Median salary = $90,572

Because the mean is higher than the median, the higher-paid professionals are pulling the average upward.

# What we can say in the project

Salary distribution is positively skewed, with most professionals concentrated in the lower-to-middle salary range and a smaller number of high earners extending beyond $300,000.

This is important because it tells us that mean salary alone doesn't tell the whole story. We'll use the median and distributions as well.

```python
salary_by_experience = (
    df.groupby("experience_level")["salary_usd"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(x=salary_by_experience.index, y=salary_by_experience.values)

plt.title("Average Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Average Salary (USD)")

plt.show()
```

There is a very clear upward relationship between career level and average salary:

Higher experience levels are associated with substantially higher average salaries in this dataset.

This is one of the strongest charts in our project, so we'll definitely use it later in the Power BI dashboard.

```python
# Calculate the average salary for each industry
salary_by_industry = (
    df.groupby("industry")["salary_usd"]
      .mean()
      .sort_values(ascending=False)
)

# Create a horizontal bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    x=salary_by_industry.values,
    y=salary_by_industry.index
)

plt.title("Average Salary by Industry")
plt.xlabel("Average Salary (USD)")
plt.ylabel("Industry")

plt.show()
```

What does the chart tell us?
- Finance has the highest average salary, approximately $111,842.
- Technology is second, at approximately $105,201.
- Energy is third, at approximately $99,754.
- Education has the lowest average salary, approximately $73,756.
The difference between Finance and Education is about $38,086.
Our finding:
Professionals in Finance have the highest average salary in this dataset, while Education has the lowest. These are differences in group averages and do not necessarily mean that industry alone determines an individual's salary.

This chart is suitable for our portfolio and future Power BI dashboard.

```python
# Calculate average salary for each education level
salary_by_education = (
    df.groupby("education_level")["salary_usd"]
      .mean()
      .sort_values(ascending=False)
)

# Create the chart
plt.figure(figsize=(10, 6))

sns.barplot(
    x=salary_by_education.values,
    y=salary_by_education.index
)

plt.title("Average Salary by Education Level")
plt.xlabel("Average Salary (USD)")
plt.ylabel("Education Level")

plt.show()
```

What does the chart tell us?
The highest average salary belongs to PhD professionals, at approximately $114,585, followed by Master's graduates at $103,030.
Self-taught and Bootcamp professionals have similar averages, at approximately $88,420 and $88,049, respectively.
Our finding is:
PhD professionals have the highest average salary in this dataset, while Bootcamp professionals have the lowest. However, education alone does not explain salary differences. Experience, industry, job title, and other factors may also matter.

Your chart looks good. We can later improve it with dollar formatting and labels showing the exact average on each bar.

```python
# Calculate AI adoption by industry
ai_adoption_by_industry = (
    df.groupby("industry")
      .agg(
          total_professionals=("uses_ai_tools_daily", "size"),
          daily_ai_users=("uses_ai_tools_daily", "sum"),
          adoption_rate=("uses_ai_tools_daily", "mean")
      )
      .reset_index()
)

# Convert adoption rate into a percentage
ai_adoption_by_industry["adoption_rate"] = (
    ai_adoption_by_industry["adoption_rate"] * 100
)

# Arrange industries from highest to lowest adoption
ai_adoption_by_industry = ai_adoption_by_industry.sort_values(
    by="adoption_rate",
    ascending=False
)

# Create the visualization
plt.figure(figsize=(10, 6))

sns.barplot(
    data=ai_adoption_by_industry,
    x="adoption_rate",
    y="industry"
)

plt.title("Daily AI Adoption Rate by Industry")
plt.xlabel("Professionals Using AI Daily (%)")
plt.ylabel("Industry")
plt.xlim(0, 100)

plt.show()
```
